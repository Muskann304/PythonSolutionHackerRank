```python
# Problem 9: Second Maximum in list
# Step 1: Read the number of inputs
n = int(input())

 # Step 2: Parse the space-separated string of scores into integers
arr = map(int, input().split())

 # Step 3: Remove duplicates by converting to a set
unique_scores = set(arr)

# Step 4: Sort the unique elements into an ascending list
sorted_scores = sorted(list(unique_scores))

# Step 5: Extract and print the runner-up (second-to-last element)
print(sorted_scores[-2])
```

```python
# Problem 10 : List
if __name__ == '__main__':
    N = int(input())
    my_list = []

    for _ in range(N):
        # Read the command line and split it into components
        # e.g., "insert 0 5" becomes ['insert', '0', '5']
        parts = input().split()
        command = parts[0]

        if command == "insert":
            index = int(parts[1])
            element = int(parts[2])
            my_list.insert(index, element)

        elif command == "print":
            print(my_list)

        elif command == "remove":
            element = int(parts[1])
            my_list.remove(element)

        elif command == "append":
            element = int(parts[1])
            my_list.append(element)

        elif command == "sort":
            my_list.sort()

        elif command == "pop":
            my_list.pop()

        elif command == "reverse":
            my_list.reverse()
```
